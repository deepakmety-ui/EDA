```python
from sklearn.tree import DecisionTreeClassifier #algorithm, for separating but before that we want to train that
```

```python
c=DecisionTreeClassifier()  #why we give fuction means it gives some works
#next we want to give the train in algoithm
c.fit(X_train,y_train)#so to fit the training data into algorithm
```

```python
#so now we can test by giving the x_test
y_pred=c.predict(X_test)
```
